```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("courtsandrinks.csv")
```

```python
# Show all columns and rows without truncation
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

# Prevent long text inside cells from being cut off
pd.set_option('display.max_colwidth', None)

# Prevent columns from wrapping onto new lines in the console
pd.set_option('display.width', 1000)
```

```python
# Count facilities in each neighborhood
facility_counts = data["neighborhood"].value_counts()

# Display the results
print("Sports facilities by neighborhood:")
print(facility_counts)

# Display the neighborhood with the most facilities
print("\nNeighborhood with the most facilities:")
print(facility_counts.idxmax())

print("Number of facilities:", facility_counts.max())
```

```python
# Only show top 15 neighborhoods
top_10 = facility_counts.head(10)

plt.figure(figsize=(10, 4))

# sort descending and plot
top_10.sort_values().plot(kind="barh", color="steelblue")

plt.title("Number of Sports Facilities by Pittsburgh Neighborhood")
plt.xlabel("Number of Facilities")
plt.ylabel("Neighborhood")

plt.tight_layout()
plt.show()
```

```python
# Group facilities by neighborhood and list unique types
sport_types = data.groupby("neighborhood")["type"].unique()

# Count the number of unique types
sport_diversity = data.groupby("neighborhood")["type"].nunique()

# Combine the counts and lists into one DataFrame
diversity_results = pd.DataFrame({
    "Diversity": sport_diversity,
    "Facility Types": sport_types
})

# Sort by greatest diversity
diversity_results = diversity_results.sort_values("Diversity", ascending=False)

# Display all neighborhoods and their facility types
display(diversity_results)

# Display the neighborhood with the greatest diversity
print("Neighborhood with the greatest diversity:")
print(diversity_results.index[0])

print("Number of different facility types:",
      diversity_results["Diversity"].iloc[0])

print("Types available:",
      diversity_results["Facility Types"].iloc[0])
```

```python
# Create a simplified sports category column
data["sport"] = data["type"].replace({
    "Basketball (Full)": "Basketball",
    "Basketball (Half)": "Basketball",
    "Street Hockey": "Land Hockey",
    "Dek Hockey": "Land Hockey"
})

# Recalculate diversity based on actual sports categories
# Group sports by neighborhood and list unique types
sport_types = data.groupby("neighborhood")["sport"].unique()

# Count the number of unique types
sport_diversity = data.groupby("neighborhood")["sport"].nunique()

# Combine the counts and lists into one DataFrame
diversity_results = pd.DataFrame({
    "Diversity": sport_diversity,
    "Facility Types": sport_types
})

# Sort by greatest diversity
diversity_results = diversity_results.sort_values("Diversity", ascending=False)

# Display all neighborhoods and their facility types
display(diversity_results)

# Display the neighborhood with the greatest diversity
print("Neighborhood with the greatest diversity:")
print(diversity_results.index[0])

print("Number of different facility types:",
      diversity_results["Diversity"].iloc[0])

print("Types available:",
      diversity_results["Facility Types"].iloc[0])
```

```python
top_10 = sport_diversity.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 4))

top_10.sort_values().plot(kind="barh", color="mediumseagreen")

plt.title("Diversity of Sports Facilities by Neighborhood")
plt.xlabel("Number of Unique Facility Types")
plt.ylabel("Neighborhood")

plt.tight_layout()
plt.show()
```

```python

# Combine quantity and diversity into one DataFrame
sportiness = pd.DataFrame({
    "Quantity": facility_counts,
    "Diversity": sport_diversity
})

# Normalize each metric to a scale of 0 to 100
sportiness["Quantity Score"] = (
    sportiness["Quantity"] /
    sportiness["Quantity"].max() * 100
)

sportiness["Diversity Score"] = (
    sportiness["Diversity"] /
    sportiness["Diversity"].max() * 100
)

# Display the normalized scores
display(sportiness)
```

```python
# Calculate the final weighted Sportiness Score
sportiness["Sportiness Score"] = (
    0.4 * sportiness["Quantity Score"] +
    0.6 * sportiness["Diversity Score"]
)

# Rank neighborhoods by their overall score
sportiness = sportiness.sort_values("Sportiness Score", ascending=False)

# Round values for easier reading
display(sportiness.round(2))

# Display the winner
print("\nMost sports-friendly neighborhood:")
print(sportiness.index[0])

print("Sportiness Score:",
      round(sportiness["Sportiness Score"].iloc[0], 2))
```

```python
# Select the 10 highest-scoring neighborhoods
top_10 = sportiness.head(10)

# Plot both normalized metrics side by side
top_10[["Quantity Score", "Diversity Score"]].sort_values("Quantity Score").plot(
    kind="barh",
    figsize=(10, 5),
    color=["steelblue", "mediumseagreen"]
)

plt.title("Sports Facility Quantity vs. Diversity")
plt.xlabel("Normalized Score (0-100)")
plt.ylabel("Neighborhood")
plt.xlim(0, 105)
plt.legend(title="Metric")

plt.tight_layout()
plt.show()
```

```python
sportiness.to_csv("sportiness_scores.csv")
```
